```python
import pandas as pd
```

# Preparation

```python
# Import traffic data
df_combined = pd.read_csv("data/combined_h1_2012_2024.csv")
df_combined.head()
```

```python
# Add day and re-arrange the columns
df_combined["Day"] = df_combined["Date"].str[-2:]
df_combined = df_combined[["Datetime", "Date", "Year", "Month", "Day", "Time", "Weekday",
                           "Location", "Location_Code",
                           "NO2", "O3", "PM10",
                           "Hr", "RainDur", "T", "WVv", "StrGlo",
                           "n_vehicles"]]
df_combined.head()
```

```python
# Filter out columns for NaN analysis
df_focused_nan = df_combined[["Year", "Month", "Day", "Time", "Weekday",
                              "Location",
                              "NO2", "O3", "PM10",
                              "Hr", "RainDur", "T", "WVv", "StrGlo",
                              "n_vehicles"]]
df_focused_nan.head()
```

# NaN overview yearly

```python
nan_overview = (
    df_focused_nan
        .groupby(["Year", "Location"])
        .agg(lambda col: col.isna().sum())
        .reset_index()
)

nan_overview
```

# NaN overview Monthly per location

```python
# Helper function for monthy overview
def monthly_nan_overview(df_original, location):
    cols_to_check = ["NO2", "O3", "PM10", "Hr", "RainDur", "T", "WVv", "StrGlo", "n_vehicles"]
    print(location)
    df_location = df_original.loc[df_original["Location"] == location]

    for year in df_location["Year"].unique():

        df_year = df_location.loc[df_location["Year"] == year]

        # NaN count per month per column
        nan_overview = (
            df_year
            .groupby("Month")[cols_to_check]
            .apply(lambda df: df.isna().sum())
        )

        # Number of unique rows with any NaN
        unique_nan_rows = (
            df_year
            .groupby("Month")[cols_to_check]
            .apply(lambda m: m.isna().any(axis=1).sum())
            .rename("Rows_with_any_NaN")
        )

        # Possible maximum NaN per month per column
        possible_max = (
            df_year
            .groupby("Month")[cols_to_check]
            .size()
            .rename("Possible_max")
        )

        # Merge all
        merged = (
            nan_overview
            .join(unique_nan_rows)
            .join(possible_max)
        )

        print(year)
        print(merged)
```

## Rosengartenstrasse

```python
# Adjust printed width to show everything in one line
pd.set_option("display.width", 200)
pd.set_option("display.max_columns", None)

monthly_nan_overview(df_focused_nan, "Zch_Rosengartenstrasse")
```
