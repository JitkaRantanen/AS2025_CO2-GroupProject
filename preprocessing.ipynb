# Traffic data

```python
import pandas as pd
```

```python
def read_files(file_paths: list, drop_columns: list):
    dfs = [
        pd.read_csv(file).assign(year=int(file[-8:-4])).drop(columns=drop_columns)
        for file in file_paths
    ]

    df = pd.concat(dfs, ignore_index=True)

    return df
```

```python
filenames_traffic = ["data/sid_dav_verkehrszaehlung_miv_od2031_2023.csv", "data/sid_dav_verkehrszaehlung_miv_od2031_2024.csv"]

df_traffic = read_files(file_paths=filenames_traffic, drop_columns=["MSID", "MSName", "ZSID", "Achse", "HNr", "Hoehe", "EKoord", "NKoord", "Richtung", "Knummer", "Kname", "AnzDetektoren", "D1ID", "D2ID", "D3ID", "D4ID", "LieferDat", "AnzFahrzeugeStatus"])
```

```python
station_map = {
    "Rosengartenstrasse (Nordstrasse)": "Zch_Rosengartenstrasse",
    "Stampfenbachstrasse (Wasserwerkstrasse)": "Zch_Stampfenbachstrasse",
    "Schimmelstrasse (Werdstrasse)": "Zch_Schimmelstrasse",
    "Zürichbergstrasse (Dreiwiesenstrasse)": "Zch_Heubeeribüel"
}
```

```python
df_traffic = df_traffic[df_traffic["ZSName"].isin(station_map)]
```

```python
df_traffic["pollutant_station"] = df_traffic["ZSName"].map(station_map)
```
